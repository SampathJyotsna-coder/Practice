```python
num = int(input("Enter a number: "))

count = 0

for i in range(1, num + 1):
    if num % i == 0:
        count += 1

if count == 2:
    print("Prime Number")
else:
    print("Not a Prime Number")
```

```python
n = int(input("Enter number of terms: "))

a = 0
b = 1

for i in range(n):
    print(a, end=" ")
    c = a + b
    a = b
    b = c
```

```python
num = int(input("Enter a number: "))
sum = 0

while num > 0:
    digit = num % 10
    sum += digit
    num = num // 10

print("Sum of digits =", sum)
```

```python
num = int(input("Enter a number: "))

temp = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

if temp == reverse:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

```python
text = input("Enter a string: ")
count = 0

for ch in text:
    if ch in "aeiouAEIOU":
        count += 1

print("Number of vowels =", count)
```

```python
numbers = [10, 25, 7, 40, 15]

maximum = numbers[0]

for num in numbers:
    if num > maximum:
        maximum = num

print("Maximum =", maximum)
```

```python
year = int(input("Enter year: "))

if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print("Leap Year")
else:
    print("Not a Leap Year")
```

matplotlib based problems

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]

plt.plot(x, y)
plt.xlabel("X values")
plt.ylabel("Y values")
plt.title("Square Numbers")
plt.show()
```

```python
import matplotlib.pyplot as plt

students = ["A", "B", "C", "D"]
marks = [80, 70, 90, 85]

plt.bar(students, marks)
plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt

expenses = [40, 30, 20, 10]
labels = ["Food", "Rent", "Travel", "Others"]

plt.pie(expenses, labels=labels, autopct="%1.1f%%")
plt.title("Monthly Expenses")
plt.show()
```

```python
import matplotlib.pyplot as plt

height = [150, 160, 170, 180, 190]
weight = [50, 55, 65, 75, 85]

plt.scatter(height, weight)
plt.xlabel("Height")
plt.ylabel("Weight")
plt.title("Height vs Weight")
plt.show()
```

```python
import matplotlib.pyplot as plt

ages = [18, 20, 20, 22, 25, 25, 25, 30, 35, 40]

plt.hist(ages)
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution")
plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = [1, 2, 3, 4, 5]
student1 = [80, 85, 78, 90, 88]
student2 = [75, 82, 80, 87, 91]

plt.plot(subjects, student1, label="Student 1")
plt.plot(subjects, student2, label="Student 2")

plt.legend()
plt.show()
```
