Cumulative Distribution Function

```python
import numpy as np
df_setosa=df[df["Species"]=="Iris-setosa"]
df_versicolor=df[df["Species"]=="Iris-versicolor"]
df_virginica=df[df["Species"]=="Iris-virginica"]
counts,bin_edges=np.histogram(df_setosa['PetalLengthCm'],bins=2,density=True)
pdf=counts/(sum(counts))
print(pdf)
print(bin_edges)
```

```python
counts,bin_edges=np.histogram(df_virginica['PetalLengthCm'],bins=10,density=True)
cdf=np.cumsum(counts)
plt.plot(bin_edges[1:],cdf)
```

```python
counts_setosa,bin_edges_setosa=np.histogram(df_setosa['PetalLengthCm'],bins=10,density=True)
cdf_setosa=np.cumsum(counts_setosa*(bin_edges_setosa[1]*bin_edges_setosa[0]))
plt.figure(figsize=(10,6))
plt.plot(bin_edges_setosa[1:],counts_setosa,label="PDF (IRIS-SETOSA)")
plt.plot(bin_edges_setosa[1:],cdf_setosa,label="CDF (IRIS-SETOSA)")
plt.title('PDF and CDF for IRIS-SETOSA')
plt.xlabel('Petal Length (cm)')
plt.ylabel('density/cumulative frequency')
plt.legend()
plt.show()
```

```python
count_setosa,bin_edges_setosa=np.histogram(df_setosa['PetalLengthCm'],bins=10,density=True)
cdf_setosa=np.cumsum(count_setosa*(bin_edges_setosa[1]-bin_edges_setosa[0]))
plt.figure(figsize=(10,6))
plt.plot(bin_edges_setosa[1:],count_setosa,label='PDF(iris-setosa petalLengthCm)')
plt.plot(bin_edges_setosa[1:],cdf_setosa,linestyle='--',label='CDF(iris-setosa petalLengthCm)')
plt.title('PDF and CDF for iris-setosa petalLengthCm')
plt.xlabel('petalLengthCm')
plt.ylabel('density/cumulative probability')
plt.legend()
plt.grid(True)
plt.show()
```

```python
species_names = df['Species'].unique()
features = ['PetalLengthCm', 'PetalWidthCm', 'SepalLengthCm', 'SepalWidthCm']
for feature in features:
  plt.figure(figsize=(10, 6))
  for species_name in species_names:
    subset = df[df['Species'] == species_name]
    counts, bin_edges = np.histogram(subset[feature], bins=10, density=True)
    bin_width = bin_edges[1] - bin_edges[0]
    plt.plot(bin_edges[1:], counts, label=f'PDF ({species_name})')
    cdf = np.cumsum(counts * bin_width)
    plt.plot(bin_edges[1:], cdf, linestyle='--', label=f'CDF ({species_name})')
  plt.title(f'PDF and CDF for {feature} by Species')
  plt.xlabel(feature)
  plt.ylabel('Density / Cumulative Probability')
  plt.legend()
  plt.grid(True)
  plt.show()
```

```python
species_names = df['Species'].unique()
features = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']
quantiles_to_calculate = [25, 50, 75, 100]

for species_name in species_names:
    print(f"\n--- {species_name} ---")
    subset = df[df['Species'] == species_name]

    for feature in features:
        print(f"  {feature}:")
        for q in quantiles_to_calculate:
            quantile_value = np.percentile(subset[feature], q)
            print(f"    {q}th Quantile: {quantile_value:.2f}")
```
