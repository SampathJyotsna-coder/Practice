Copies and views

```python
import numpy as np
a=np.arange(10)
print(a)
b=a[::2]
print(b)
print(np.shares_memory(a,b))
b[0]=10
print(b)
print(a)
a=np.arange(10)
c=a[::2].copy()
print(c)
print(np.shares_memory(a,c))
c[0]=10
print(c)
print(a)
```

Numerical Operations

Elementwise operation

```python
import numpy as np
a=np.array([1,2,3,4])
b=np.ones(4)+1
print(a)
print(b)
print(a-b)
print(a+b)
print(a*b)
print(a/b)
```

Comparisions

```python
a=np.array([1,2,3,4])
b=np.array([5,2,2,4])
print(a==b)
print(a>b)
a=np.array([1,2,3,4])
b=np.array([5,2,2,4])
c=np.array([1,2,3,4])
print(np.array_equal(a,b))
print(np.array_equal(a,c))
```

Logical operations

```python
a=np.array([1,1,0,0],dtype=bool)
b=np.array([1,0,1,0],dtype=bool)
print(np.logical_or(a,b))
print(np.logical_and(a,b))
np.logical_and(a,b)
```
